# Titanic Dataset - Exploratory Data Analysis

This notebook performs a comprehensive exploratory data analysis (EDA) of the Titanic dataset, including data loading, visualization, and relationship analysis to understand the factors influencing survival.

## Section 1: Import Required Libraries

Import pandas, numpy, matplotlib.pyplot, and seaborn for data manipulation and visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import xlabel
from matplotlib.pyplot import ylabel
from matplotlib.pyplot import show
from matplotlib.pyplot import title
import seaborn as sns
```

## Section 2: Load and Preview Dataset

Load the train.csv file using pandas and display the first few rows to understand the dataset structure.

```python
# Load dataset
df = pd.read_csv("train.csv")

# Basic view
df.head()
```

## Section 3: Dataset Information and Statistics

Use df.info() to display data types and non-null counts, and df.describe() for statistical summary of numeric columns.

```python
# Dataset Info
df.info()
```

```python
# Statistical Summary
df.describe()
```

## Section 4: Categorical Distribution Analysis

Use value_counts() to analyze the distribution of Survived, Pclass, and Sex columns.

```python
# Categorical Distribution
print("Survival Distribution:")
print(df['Survived'].value_counts())
print("\nPassenger Class Distribution:")
print(df['Pclass'].value_counts())
print("\nGender Distribution:")
print(df['Sex'].value_counts())
```

## Section 5: Survival Visualization

Create a count plot using seaborn to visualize the distribution of survivors vs non-survivors.

```python
# Survival Count Plot
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()
```

## Section 6: Gender vs Survival Analysis

Create a count plot showing the relationship between gender and survival with hue differentiation.

```python
# Gender vs Survival
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Gender vs Survival")
plt.show()
```

## Section 7: Passenger Class vs Survival Analysis

Create a count plot to visualize how passenger class affects survival outcomes.

```python
# Passenger Class vs Survival
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Passenger Class vs Survival")
plt.show()
```

## Section 8: Age Distribution Analysis

Create a histogram with kernel density estimation (KDE) to visualize the age distribution of passengers.

```python
# Age Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], kde=True)
plt.title("Age Distribution")
plt.show()
```

## Section 9: Fare Distribution Analysis

Create a histogram with KDE to visualize the fare distribution across all passengers.

```python
# Fare Distribution
sns.histplot(df['Fare'], kde=True)
plt.title("Fare Distribution")
plt.show()
```

## Section 10: Age vs Survival Relationship

Create a box plot to compare age distributions between survivors and non-survivors.

```python
# Age vs Survival
sns.boxplot(x='Survived', y='Age', data=df)
plt.title("Age vs Survival")
plt.show()
```

## Section 11: Fare vs Survival Relationship

Create a box plot to compare fare distributions between survivors and non-survivors.

```python
# Fare vs Survival
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title("Fare vs Survival")
plt.show()
```

## Section 12: Correlation Analysis

Calculate numeric correlations and visualize using a heatmap with annotations.

```python
# Correlation Heatmap
plt.figure(figsize=(8,6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## Section 13: Pairwise Relationships Overview

Create a pairplot to visualize relationships between Survived, Pclass, Age, and Fare with hue by survival status.

```python
# PairPlot Relationship Overview
sns.pairplot(df[['Survived','Pclass','Age','Fare']], hue='Survived')
plt.show()
```

## Section 14: Missing Values Assessment

Use isnull().sum() to identify and display the count of missing values in each column.

```python
# Missing values
print("Missing Values Count:")
print(df.isnull().sum())
```

## Summary

This exploratory data analysis has provided insights into the Titanic dataset including:
- Data types, structure, and missing values
- Distribution of survival outcomes
- Impact of gender and passenger class on survival
- Age and fare distributions
- Correlations between numeric features
- Relationships between key variables and survival status

These insights can be used for further data preprocessing and predictive modeling.
