Step: 1 - Create Database and Insert Sample Data

```python
import sqlite3
```

# Connect to SQLite database (this will create the file if it doesn't exist)

```python
conn = sqlite3.connect("sales_data.db")
cursor = conn.cursor()
```

# Create sales table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    product TEXT,
    quantity INTEGER,
    price REAL
)
""")
```

# Insert sample data

```python
sales_data = [
    ("Laptop", 5, 60000),
    ("Laptop", 3, 62000),
    ("Phone", 10, 20000),
    ("Phone", 7, 21000),
    ("Tablet", 6, 30000),
    ("Tablet", 4, 28000)
]

cursor.executemany("INSERT INTO sales (product, quantity, price) VALUES (?, ?, ?)", sales_data)

# Commit and close
conn.commit()
conn.close()

print("Database and table created successfully!")
```

Step: 2 - Connect to Database & Run SQL Query

```python
import sqlite3
import pandas as pd
```

conn = sqlite3.connect("sales_data.db")

# SQL Query
query = """
SELECT 
    product,
    SUM(quantity) AS total_qty,
    SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
"""


```python
# Load into pandas DataFrame
df = pd.read_sql_query(query, conn)

conn.close()

print("Sales Summary:")
print(df)
```

Step: 3 - Create Basic Bar Chart (Matplotlib Only)

```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
plt.bar(df["product"], df["revenue"])
```

```python
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
```

```python
plt.show()
```
